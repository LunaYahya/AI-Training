## Array Attributes, Indexing & Slicing
(By using this rule array[row,column])


```python
m=np.array([[1,2,3],[5,6,7],[8,9,10]])
print(m.shape)
```

```python
print(m[1,2]) #row:1,col:3
```

```python
print(m[:,2]) # all rows, col:2
```

```python
print(m[2,:]) #row:1,all col
```

## Vectorized Operations
apply operations on arrays using numpy to make faster and more readable code than using loops 

```python
array3=np.array([1,10,15,30])
sum=(array3+array3) #sum element by element
print(sum)
```

```python
mult=(array3*array3) #multiply element by element
print(mult)
```

```python
m=(array3.mean()) #find the mean
print(m)
```

```python
mx=(array3.max()) #find the max element
print(mx)
```

```python
mn=(array3.min()) #find the min element
print(mn)
```

```python
a=(array3[array3>3]) #boolean masking
print(a) 
```

## Broadcasting
Numpy automatically expands the smaller array ro match the larger one then doing the operations
for example adding a single number to every element,or adding a 1D array to 2D array


```python
m=np.array([[1,2,3],[4,5,6]])
r=np.array([5,10,20])
print(m+r)
```

```python
print(m+5)
print(r+7)
```

## Hands-On Lab

Step 1: Create a 2D array of shape (4, 4) filled with values 1-16 and print its shape and dtype.


```python
arr=np.arange(1,17,1)
arr1=arr.reshape(4,4)
print(arr1)
print(arr1.shape)
print(arr1.dtype)
```

Step 2: Use slicing to extract the second column and the last row.


```python
print(arr1[:,1]) #the second colum
```

```python
print(arr1[-1]) #the last row
```

Step 3: Use a boolean mask to select all values greater than the array's mean.


```python
mean1=arr1.mean()
print("values greater than the mean")
print(arr1[arr1>mean1])
```
