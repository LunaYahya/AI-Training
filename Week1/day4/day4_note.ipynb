## Cleaning Data
remove duplicates value,wrong data types or missing values

```python
df.isnull().sum() #number of null values
```

```python
df["Age"].fillna(df["Age"].mean(),inplace=True)  #fill null values with mean of age
```

```python
df.dropna(inplace=True) #drop rows with missing value
```

```python
df.drop_duplicates(inplace=True) #remove duplicates from the dataset
```

## Grouping and Aggregation
Groups data by a column and computes statistics for each group

```python
df.groupby("Passengerid") ["Age"].mean()
```

```python
df.groupby("Passengerid").agg({"Age":"mean","Fare":"max"})
```
