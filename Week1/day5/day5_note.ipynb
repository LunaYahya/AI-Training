## Step 2: Use NumPy to compute at least one derived numeric feature or summary statistic.

```python
max_math=np.max(df["math score"])
print("Highest math score is:",max_math)
```

```python
mean_math=np.mean(df["math score"])
print("Mean math score is:",mean_math)
```

```python
df["Total_Score"]=np.sum(df[["math score","reading score","writing score"]],axis=1)
df.head()
```

## using numpy a new column a named "Total_Score" was added by sum math,reading,writing score for each student

## Step 3: Produce at least three labeled plots (including a histogram and a scatter plot) exploring the data.

```python
plt.hist(df["math score"])
plt.title("Distribution of math srocres")
plt.xlabel("Math socre")
plt.ylabel("Number of students")
plt.show()
```

## Observation
the histogram shows the distribution of math scores among students,Most of students score between 60 and 80.

```python
plt.scatter(df["reading score"],df["math score"])
plt.title("Reading score vs math score")
plt.xlabel("Reading Score")
plt.ylabel("Math Score")
plt.show()
```

## Observation
The scatter plot shows a positive linear realtionship between reading and math scores
(students who score higher in reading tend to also score higher in math)

```python
math_score=df["math score"].head(15)
plt.plot(math_score.values,color='red')
plt.title("Math scores for  first 15 students")
plt.xlabel("student Index")
plt.ylabel("Math Score")
plt.show()
```

## Observation
This line chart shows the changes in math scores accross the first 15 students.
