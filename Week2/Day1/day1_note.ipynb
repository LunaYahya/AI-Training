##  Descriptive Statistics
Descriptive Statistics is the part of statistics that is 
utilized for summarizing,organizing,and describing important 
characteistics of a dataset.The proccess of getting insight into
the information through descriptive statistics involves the use
of central tendency measures (mean,median,mode) and measures of
variabilliy(range,variance,standard deviation,IQR).

## Measures of Central Tendency
## (Mean),(Median),(Mode)

## Mean
we can find the mean by adding all the values in the dataset,then divide vy the total numbers of values.

## Median
after arrange the data in ascending order if the number is even
the median is the average of the two middle values and if the number is odd the median is the middle value.

## Mode
It's the value that appears most frequently in the dataset.It can be one or more or none.

```python
import numpy as np
```

```python
data=np.array([20,12,15,12,40])
print(np.mean(data))
print(np.median(data))
```

## Measures of Spread
## (Range),(Variance),(Standard deviation),(Interquartile range(IQR)).

## Range
maximum value-minimum value

## Variance
indicates how far data values are from the mean.It's computed by taking the average of the sequares of the differences of the values from the mean.

## Standard deviation (SD)
indicates the spread of data around the mean. It is equal to the sequare root of the variance.If the (SD) is small the data is clustered around the mean.On the other hand, a large (SD) implies that data is widely spread.

## interquartile range (IQR)
indicates the spread of the middle 50% of the data. The steps for computation are as follow:
1- arrange the data increasing order
2- compute Q1 (first quartile)
3- compute Q3 (third quartile)
4- IQR=(Q3-Q1) it is not sensitive to outliers like the range.

```python
data2=np.array([10,20,40,12,100])
print(np.std(data2))
print(np.var(data2))
q1,q3=np.percentile(data2,[25,75])
print(q3-q1)
```

## Hands-On Lab: Describing a Real Dataset

```python
import pandas as pd
```

```python
df=pd.read_csv("tested.csv")
df.info()
df.head()
```

```python
data3=df["Fare"].dropna()
mean=np.mean(data3)
print("mean:",mean)
median=np.median(data3)
print("median:",median)
mode=data3.mode()[0]
print("Mode:",mode)
sd=np.std(data3)
print("standard deviation:",sd)
Q1=np.percentile(data3,25)
Q3=np.percentile(data3,75)
iqr=Q3-Q1
print("IQR:",iqr)
```

## Wich measure better represents a  typical value?
The median provides a better representation of the typical value
for the Fare since there are outliers that make the average higher. The median is not influenced by outliers and therefor provides an accurate representation of the typical value.
