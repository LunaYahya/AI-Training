## Probability & Distributions
The term probabillity can be defined as the measure of accurrence of an event in mathematics.It is on of the basic concepts in Machine Learning as many ML algorithms give probabillistic predicitions insted of definite answers.

## Probabillity Basics
Probabillity measures how likely an event is to happen
it is value ranges from 0(impossible) to 1 (certain)

```python
#probabillity of collecting a green ball
green_balls=3
red_balls=7
total_balls=10
probabillity_of_green=red_balls/total_balls
print("probabillity of collecting a green ball is",probabillity_of_green)
```

```python
#probabillity of drawing a green ball
green=17
cards=70
probabillity=green/cards
print("probabillity of drawing a green ball is",probabillity)
```

The probabillity is calculated by dividing the number of favorable outcomes by the total numbers of possible outcomes.

## Core Rules
## 1- Complement Rule
The complement Rule provides the probabillity of an event not happening
## Formula
P(not A)=1-P(A)

```python
#probabillity of passing an exam
passing_exam=0.74
probabillity_of_fail=1-passing_exam
print("probabillity of falling",probabillity_of_fail)
```

Computes the complement probability by subtracting the probability of success from 1

## Addition Rule
The addition rule calculates the probability of occurrence of one of the two events.
## Formula
P(A or B) = P(A) + P(B) - P(A and B)

```python
probability_of_math=0.40
probability_of_programming=0.30
probability_of_both=0.20
probability_of_math_or_programming=probability_of_math+probability_of_programming-probability_of_both
print(probability_of_math_or_programming)
```

Computes the union of two events by adding their probabilities and subtracting the intersection of the two events.

## Multiplication Rule
If two events are independent, the probability of the occurrence of both is equal to the product of their probabilities.
## Formula
P(A and B) = P(A) x P(B)

```python
probability_of_head=0.5
probability_of_five=1/6
probability_of_both=probability_of_head*probability_of_five
print(probability_of_both)
```

Computes the probability that two independent events happen at once by multiplying their probabilities.

## Conditional Probabillity
probabillity of an events occurrence knowing that another event has already occurred.
## Formula
P(A|B)=P(A and B)/P(B)

```python
female_students=30
female_DR=10
P=female_students/female_DR
print(P)
```

Performs simple division to compute the ratio of female students to female doctors. 

## Bayes' Theorem
the Bayes' Theorem is applied in updating the probabillities of events when new pieces of evidence are acquired, it forms the basic of Naive Bayes technique used in Machine Learning for classification purposes such as spam and medical diagnosis.
## Formula
P(A|B)=(P(B|A)*P(A))/P(B)
Bayes' Theorem combines:
-Pior: initial belief befor seeing evidence.
-likelihood:probabillity of the evidence if the event is true
-Posterior:Updating probabillity after considering the evidence.

```python
p_disease=0.01
p_positive_given_disease=0.98
p_positive=0.04
p_disease_given_positive=(p_positive_given_disease*p_disease)/p_positive
print(p_disease_given_positive)
```

Uses Bayes theorem to compute the inverse conditional probability of disease given the test result is positive.

## Common Distributions
Probability distribution is defined as the distribution of the values in a data set.
## Normal (Gaussian)
- Symmetric and bell shaped.
- The majority of values lie around the mean.
- Widely seen in nature; for example, height and measurement errors.


```python
import numpy as np
data=np.random.normal(50,10,100)
print("Mean:",np.mean(data))
print("SD",np.std(data))
```

Draws 100 random numbers from normal distribution with mean=50, std=10 and computes their sample mean and sample sd.

## Binomial Distribution
- Used to describe the number of successes in a given number of independent trials.
- Example: Number of heads in 10 flips of a coin.

```python
passed=np.random.binomial(n=20,p=0.7)
print(passed)
```

Computes the number of successes in 20 binomial trials with success probability per trial equal to 0.7.


## Uniform Distribution
- Equally likely outcome for each event.
- Example: Tossing a fair six-faced die.

```python
temp=np.random.uniform(15,40,10)
print(temp)
```

Computes 10 uniformly distributed data samples between minimum and maximum values of 15 and 40 respectively.
