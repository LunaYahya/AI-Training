## Linear Algebra for ML
Linear Algebra is a branch of math that covers vectors,matrices,and linear transformations.it forms the basis of how data can be analyzed and represented and is thus crucial in Machine Learning.

## Why Linear Algebra Matters
Linear Algebra forms the basis of Machine Learning.
- A vector is a single data point.
- A matrix is an entrie dataset.
- Machine Learning algorithms perform dot products to predict values.
- Multipilication of matrices enables predicition on many data points at once.

## Matrices
A matrix is a group of vectors,row crossepond to samples and columns crossepond to features.

```python
students=np.array([[25,100,50],[50,60,70],[100,200,400]])
print(students)
print(students.shape)
```

3 rows means 3 samples

## Dot Product
the dot product entails multiplying crossponding entries in two vectors and summing the products,to generates a single number,it used in Machine Learning to generate predicitions. 

```python
features=np.array([1,2,3])
weights=np.array([4,5,6])
prediction=np.dot(features,weights)
print(prediction)
```

By hand
(1*4)+(2*5)+(3*6)
=(4)+(10)+(18)
=32

```python
x=np.array([2,5])
y=np.array([3,6])
result=np.dot(x,y)
print(result)
```

## Matrix Multiplication
matrix multiplication involoves performing dot products on multiple instance simultaneously
## Rule
(mxn)x(nxp)
the inner dimensions should match.

```python
w=np.array([[1,2,3],[4,5,6]])
z=np.array([2,3,1])
predict=w@z
print(predict)
```

By hand
=((1*2)+(2*3)+(3*1)),((4*2)+(5*3)+(6*1))
=(2+6+3),(8+15+6)
=(11,29)

## Compatibillity Rule for Matrices
in order to multiply matrices, the number of columns of the first matrix needs to match the number of rows of the second matrix otherwise the NumPy will throw a ValueError
