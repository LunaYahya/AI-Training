Expolartory Data analysis (EDA): is analysis of a dataset in order to find out about its underlying structure ,patterns,outliers and make it ready for analysis and machine learning.

## correlation
correlation refers to the degree of relationship between two quatitiative variables.The value of it ranges between -1 and +1.

## import libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("student_performance_dataset.csv")
df.head(10)
```

```python
df.describe()
```

## private analysis
private analysis is the analysis that takes place between two variables.
The scatter plot shows the relationship between two numerical variables while the grouped box plot shows the difference among categories of a numerical variable.

```python
sns.scatterplot(data=df,x="study_time_hours",y="final_exam_score")
plt.title("relationship between study time hours and final exam score")
plt.show()
```

this plot exploress wether sleep duration is have a relationship with exam performance.

```python
sns.boxplot(data=df,x="previous_grade",y="final_exam_score")
plt.title("relationship between previous grade and final exam score ")
plt.show()
```

this plot shows that better previous grades lead to better exam results.

```python
corr=df.corr(numeric_only=True)
print(corr)
```

```python
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.show()
```

The following heatmap shows the correlation between the numerical variables where values close to 1 is suggst positive correlations and those close to -1 denot negative correlaations.
In short it shows the strength of relationships between variables.

```python
sns.pairplot(df[["study_time_hours","previous_grade","final_exam_score"]])
plt.show()
```

Pair plot help us get a quick idea about  numerical relationships.

## Step 3: Identify the two or three strongest relationships and state what they might mean for a future model.

From the correlation analysis the following strongest relationships in the dataset include:
- previous grade and final exam score: the students who performed well in their previous grade also score highly in their final exam, so this mean that the previous grades can help to predict future performance.
- study time hours and final exam score: the students who take longer time to study for the exam perform better than who take less time, so study hours are a factor to consider in the prediction of the performance.
- attendace percent and final exam score: the students with high attendance record have high final exam scores,therfor the attendance will play a role in the predicting the students performance.
