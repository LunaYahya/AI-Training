## Supervised Learning Concepts & the Scikit-learn API
Supervised Learning is a form of machine learning in wich model learns from labeled data to predict the right output for future input,in supervised learning the features(x) are used to learn the patterns in order to predict the target(y).

## Types of Supervised Learning
There are two major types of supervised learning:
- Regression: predicts numerical outputs.
## Examples:
- salary
- final exam score
- house price
- Classification: predicts categories or classes.
## Examples:
- spam / not spam
- pass / fail

Typical supervised learning process consists of feature and target selection,data partitioning,into train and test sets,training of the model,prediction,and model evaluation

### Hands-on-lab:
## Step 1: Load a provided dataset and separate it into features X and target y.

```python
x=df.drop("final_exam_score",axis=1)
y=df["final_exam_score"]
print(x.head())
print(y.head())
print("Number of Features:",x.shape[1]) ## To knows the number of features.
print("Number of Samples:",x.shape[0]) ## To Knows the number of samples.
```

## Why do we have to split the dataset?
when we are using the same dataset to evaluate and train our model,then there will be chances that the model will
memorize the data and not learn anything from it,and using separate test data we can determine how good our model performs for new data.

## Step 2: Perform an 80/20 train/test split with a fixed random_state.


```python
x_train,x_test,y_train,y_test=train_test_split(
    x,y,test_size=0.2,random_state=42
)
```

## Step 3: Confirm the shapes of X_train, X_test, y_train, y_test are consistent and print them.

```python
print("x_train:",x_train.shape)
print("x_test:",x_test.shape)
print("y_train:",y_train.shape)
print("y_test:",y_test.shape)
```

```python
print(len(x_train)==len(y_train))
print(len(x_test)==len(y_test))
```

## Confirming the split consistency
it is confirmed that the number of samples in x_train is the same number of lables in y_train and the number of samples in x_test is the same number of lables in y_test.
This is very necessary because each row of features has to be paired with exactly one target variable,the inconsistency will cause the learning proccess to be faulty.
