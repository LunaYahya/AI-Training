## Final Model Serving Contract

The deployed model follows a fixed serving contract.

| Component | Specification |
|---|---|
| Task | Melanoma skin-lesion classification |
| Classes | Benign / Malignant |
| Input | RGB image |
| Image size | 128 × 128 |
| Channels | 3 |
| Image reader | OpenCV |
| Color conversion | BGR → RGB |
| Normalization | Pixel values divided by 255.0 |
| Model format | Keras `.keras` |
| Model artifact | `melanoma_cnn.keras` |
| Metadata | `model_metadata.json` |
| Positive class | Malignant |
| Reference threshold | 0.50 |
| Selected threshold | 0.35 |
| Reproducibility seed | 42 |

The selected threshold of 0.35 is used by the deployed application.
The decision rule is:

probability_malignant >= 0.35 → Malignant

Otherwise:

probability_malignant < 0.35 → Benign

## Final Evaluation Results

The Week 8 evaluation was performed on a balanced test set containing
2,000 images:

- 1,000 Benign
- 1,000 Malignant

At the reference threshold of 0.50, the reported results were:

| Metric | Value |
|---|---:|
| Accuracy | 0.8755 |
| Malignant Precision | 0.9243 |
| Malignant Recall | 0.8180 |
| Malignant F1-score | 0.8679 |
| ROC-AUC | 0.9517 |
| PR-AUC | 0.9445 |
| False Positives | 67 |
| False Negatives | 182 |

Threshold analysis identified 0.35 as the selected deployment threshold.

At threshold 0.35:

| Metric | Value |
|---|---:|
| Accuracy | 0.8915 |
| Malignant Precision | 0.8746 |
| Malignant Recall | 0.9140 |
| Malignant F1-score | 0.8939 |
| ROC-AUC | 0.9517 |
| PR-AUC | 0.9445 |
| False Positives | 131 |
| False Negatives | 86 |

The threshold change does not change ROC-AUC or PR-AUC because these
metrics evaluate the ranking of prediction scores rather than a single
classification threshold.

The selected threshold increases malignant recall and reduces false
negatives compared with the 0.50 reference threshold, while increasing
false positives.

##  Project Limitations

The current system has several limitations.

First, the model is trained and evaluated using a specific dataset, so
its performance may not generalize to every real-world population,
camera, imaging condition, or clinical setting.

Second, the model can still make both false-positive and false-negative
predictions.

Third, changing the classification threshold changes the balance between
false positives and false negatives. The selected threshold of 0.35
improves malignant recall compared with the 0.50 reference threshold,
but it also increases false positives.
